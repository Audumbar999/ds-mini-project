```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('cleaned_deliveries.csv')
plt.figure(figsize=(8, 5))
sns.histplot(df['final_score'], bins=25, kde=True, color='skyblue')
plt.title('Distribution of 1st Innings Final Scores in IPL')
plt.xlabel('Final Score')
plt.ylabel('Match Frequency')
plt.grid(True, linestyle='--', alpha=0.6)
plt.savefig('score_distribution.png')
plt.show()
plt.figure(figsize=(8, 5))
sns.scatterplot(x='current_runs', y='final_score', hue='wickets_fallen', data=df, palette='viridis', alpha=0.5)
plt.title('Current Runs vs Final Score (Colored by Wickets Fallen)')
plt.xlabel('Current Runs')
plt.ylabel('Final Score')
plt.grid(True, linestyle='--', alpha=0.6)
plt.savefig('runs_vs_final_score.png')
plt.show()
plt.figure(figsize=(6, 4))
numeric_cols = df[['current_overs', 'current_runs', 'wickets_fallen', 'final_score']]
sns.heatmap(numeric_cols.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Feature Correlation Heatmap')
plt.tight_layout()
plt.savefig('correlation_heatmap.png')
plt.show()
print("EDA analysis completed and charts generated successfully!")
```
