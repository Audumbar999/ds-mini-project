```python
import pandas as pd
```

```python
df = pd.read_csv('deliveries.csv')
```

```python
df_first = df[df['inning'] == 1].copy()
```

```python
df_first['total_runs'] = df_first['total_runs'].fillna(0)
df_first['is_wicket'] = df_first['player_dismissed'].notnull().astype(int)
```

```python
df_first['current_runs'] = df_first.groupby('match_id')['total_runs'].cumsum()
df_first['wickets_fallen'] = df_first.groupby('match_id')['is_wicket'].cumsum()
```

```python
df_first['ball_count'] = df_first.groupby('match_id').cumcount() + 1
df_first['current_overs'] = (df_first['ball_count'] / 6.0).round(2)
```

```python
final_scores = df_first.groupby('match_id')['total_runs'].sum().reset_index()
final_scores.rename(columns={'total_runs': 'final_score'}, inplace=True)
df_first = df_first.merge(final_scores, on='match_id')
```

```python
cleaned_data = df_first[df_first['current_overs'] >= 5.0][ ['match_id', 'batting_team', 'bowling_team', 'current_overs', 'current_runs', 'wickets_fallen', 'final_score'] ]
```

```python
print("Missing values check:\n", cleaned_data.isnull().sum())
```

```python
cleaned_data.to_csv('cleaned_deliveries.csv', index=False)
print("Data cleaned successfully! Total rows ready for modeling:", len(cleaned_data))
```
